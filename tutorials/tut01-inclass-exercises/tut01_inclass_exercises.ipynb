## 1 · Write a function

Write `compute_s_n(n)` that returns the sum of squares:

$$S_n = 1^2 + 2^2 + 3^2 + \dots + n^2$$

Then report the value for `n = 100`.

````r
# YOUR CODE
# compute_s_n <- function(n) { ... }
# compute_s_n(100)
````

<details>
<summary><b>Reveal answer</b></summary>

```r
compute_s_n <- function(n) {
  s <- 0
  for (i in 1:n) {
    s <- s + i^2
  }
  return(s)
}

compute_s_n(100)   # 338350
```

The for-loop version works because R's `1:n` makes a sequence. Note: `s <- s + i^2` accumulates. (vectorized alt: `sum((1:100)^2)`)
</details>

**d) Filter to mpg > 20** → `mtcars_filtered` (use `filter`).

````r
# YOUR CODE
# mtcars_filtered <- mtcars |> filter(?)
# head(mtcars_filtered)
````

<details>
<summary><b>Reveal answer</b></summary>

```r

head(mtcars_filtered)
```

`filter` keeps rows where the condition is TRUE. Note this filters the whole row, not columns — that's `select`'s job.
</details>

**e) Select only `mpg` and `cyl`** from the filtered set → `mtcars_selected` (use `select`).

````r
# YOUR CODE
# mtcars_selected <- mtcars_filtered |> select(?)
# head(mtcars_selected)
````

<details>
<summary><b>Reveal answer</b></summary>

```r
mtcars_selected <- mtcars_filtered |> select(mpg, cyl)
head(mtcars_selected)
```

`select` picks **columns** (keeps all rows). `filter` picks **rows** (keeps all columns). Easy to mix up — remember, filter = rows, select = columns.
</details>

**f) Group by `cyl`, summarize mean mpg** → `mtcars_grouped` (use `group_by` + `summarize`).

````r
# YOUR CODE
# mtcars_grouped <- mtcars_selected |> group_by(cyl) |> summarize(?)
# mtcars_grouped
````

<details>
<summary><b>Reveal answer</b></summary>

```r
mtcars_grouped <- mtcars_selected |> group_by(cyl) |> summarize(mean_mpg = mean(mpg))
mtcars_grouped
```

`group_by(cyl)` splits data by cylinder value; `summarize(mean_mpg = mean(mpg))` aggregates each group to one row. Result: one mean mpg per cylinder group. This is the classic **split-apply-combine** pattern — the heart of data analysis.
</details>

## Recap — the 5 dplyr verbs + pipe

| verb | does | rows/cols |
|------|------|-----------|
| `filter` | keep rows matching a condition | rows |
| `select` | keep chosen columns | cols |
| `mutate` | add / overwrite a column | cols |
| `arrange` | sort rows | rows |
| `summarize` (with `group_by`) | collapse groups to summaries | rows ↓ |

The pipe `|>` threads the left side as the first argument to the right side:
`df |> filter(...) |> select(...)` = `select(filter(df, ...), ...)`. It makes the chain read top-to-bottom like a recipe.

**Quick self-quiz:**  
- `filter` vs `select`: which drops columns, which drops rows?
- What does `mutate` do that `summarize` doesn't?
- What does `group_by` *alone* do to a data frame? (It doesn't change the rows you see — it just remembers the grouping for the next verb.)
