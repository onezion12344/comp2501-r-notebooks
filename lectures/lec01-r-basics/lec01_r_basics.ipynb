## The big picture: why R

The lecture makes a few key points — read them, they'll save you confusion later:

- **R is not a "software engineering" language.** It was built by *statisticians* as an interactive environment for data analysis. It won't follow the conventions you know from C/Java/Python — don't fight it.
- **Interactivity is the point.** You explore data quickly. Hence the console, not just scripts.
- **You can save your work as scripts** — the key feature that makes analysis *reproducible*.
- **R's superpower is visualization**, ahead of Python (pandas). We'll get there.

**The running joke in the slides:** R looks easy to learn (`x <- 1`) but subtle to master (integer vs numeric, vectorization, recycling). This notebook only covers the easy half — but the *why* behind it.

## 1 · Objects & assignment

In R, a variable is called an **object**. You assign with `<-` (a left arrow), **not** `=`.

```r
a <- 1   # assign 1 to object 'a'
b <- 1
c <- -1
```

This solves the quadratic formula example from the lecture: `x^2 + x - 1 = 0` → `x = (-b ± √(b²-4ac)) / 2a`.

**Why `<-` and not `=`?**  
- `=` also works in R, but is **not recommended** — it first means "give" in some contexts and can confuse.
- `<-` reads as **"assign to"** and makes intent unambiguous.
- When you see `x == 1`, that's **comparison** (equal). `x <- 1` is **assignment**. Different arrows, different meaning.

### Remove an object, and a subtlety: case-sensitivity

- `rm(a)` removes object `a`. After that, `a` gives an error: `object 'a' not found`.
- **R is case-sensitive.** `True` (uppercase T) is **not** the logical value — it's an undefined object → error. The logical values in R are `TRUE` and `FALSE` (or `T`/`F`).

## 3 · Functions

R is a **functional** language — almost everything is a function call. Two things matter most:

**a) Arguments, and their defaults.** Look at `log`:

```r
log(10)        # [1] 2.302585  (natural log, base e)
log(8, 2)      # [1] 3        (base 2)
log(x = 8, base = 2)
log(base = 2, x = 8)   # named args → order doesn't matter
```

The help page shows `log(x, base = exp(1))`. So `x` is **required**; `base` is **optional** with a default. Named arguments let you skip the order.

**b) Looking things up.** `?log` (or `help("log")`) opens the docs. Try `?"+"` and `?"^"` for operators. (Bonus: `?l"og without parens` shows how typing a built-in function name alone just echoes its definition.)

**Task 2 — predict, then confirm.** Before running, predict the output of each line:

````r
log(100)          # predict: ??
log(100, 10)     # predict: ??
log(2, 8)        # predict: ??  (careful — is this log_2(8) or log_8(2)?)
log(base = 2, x = 8)
````

<details>
<summary><b>Reveal answer</b></summary>

```r
log(100)          # [1] 4.60517  (natural log)
log(100, 10)     # [1] 2        (log base 10)
log(2, 8)        # [1] 0.3333   (this is log_8(2) = 1/3 — arg order = f(x, base))
log(base = 2, x = 8)   # [1] 3   (now x=8, base=2 → log_2(8)=3)
```

The third line is the classic gotcha: **positional arguments** go in the order `(x, base)`, so `log(2, 8)` means log with x=2, base=8. If you meant log base 2 of 8, you must either spell out `log(base=2, x=8)` or use `log(8, 2)`.
</details>

## 4 · Atomic data types

Use `class()` to see an object's type. The major atomic types in R:

| type | meaning | example | `class()` result |
|------|---------|---------|------------------|
| `numeric` | float/double | `3` | `"numeric"` |
| `integer` | whole number | `3L` (note the L) | `"integer"` |
| `logical` | boolean | `TRUE` | `"logical"` |
| `character` | string | `"3"` | `"character"` |
| `factor` | categorical | (later) | `"factor"` |

**Task 3 — classify.** Predict each `class()` result, then run:

````r
class(3)
class(3L)
class(TRUE)
class("3")
class(3.0)

# and the .is. family:
is.numeric(3)
is.integer(3)
is.integer(3L)
is.character("3")
is.logical(TRUE)
````

<details>
<summary><b>Reveal answer</b></summary>

```r
class(3)       # [1] "numeric"
class(3L)      # [1] "integer"
class(TRUE)    # [1] "logical"
class("3")    # [1] "character"
class(3.0)     # [1] "numeric"   <- even "3.0" is a double!

is.numeric(3)      # TRUE
is.integer(3)      # FALSE  (3 is a double, not integer — the lecture's subtle point)
is.integer(3L)     # TRUE
is.character("3") # TRUE
is.logical(TRUE)   # TRUE
```

The key insight: `3` and `3L` are **different** types. `3` is a `numeric`/double; `3L` is `integer`. R auto-coerces in most cases, but the difference matters for: indexing, memory, and why `is.integer(3)` is `FALSE`.
</details>

````r
# Q2 — the vectorization question
v <- c(1, 2, 3, 4, 5)
v * 2
v + c(10, 20)
# what does R do when the lengths don't match? (predict, then run)
````

<details>
<summary><b>Q2 answer</b></summary>

```r
v * 2             # [1]  2  4  6  8 10   (elementwise multiply — no loop needed!)
v + c(10, 20)     # [1] 11 22 13 24 15   (RECYCLING: R recycles the shorter vector 10,20 → 10,20,10,20,10)
```

Two huge R ideas here:
- **Vectorization**: you operate on the *whole* vector at once, no `for` loop. This is why R is fast.
- **Recycling**: when lengths differ, R repeats the shorter one. Great when intended (b + 1), dangerous when not.
